```python
#importamos el data set
import pandas as pd

ruta = "data/raw/Online Retail.xlsx"

df = pd.read_excel(ruta)
df.head()
```

```python
# Tamaño, columnas, tipos y valores no nulos
print(df.shape)
df.info()
```

```python
# Valores faltantes
pd.DataFrame({
    "faltantes": df.isna().sum(),
    "porcentaje": df.isna().mean() * 100
}).sort_values("faltantes", ascending=False)
```

```python
# Filas exactamente duplicadas
df.duplicated().sum()
```

```python
# Rangos y valores extremos
df[["Quantity", "UnitPrice"]].describe()
```

```python
# Situaciones que debemos investigar
print("Cancelaciones:", df["InvoiceNo"].astype(str).str.startswith("C").sum())
print("Cantidad cero o negativa:", (df["Quantity"] <= 0).sum())
print("Precio cero o negativo:", (df["UnitPrice"] <= 0).sum())
print("Fechas:", df["InvoiceDate"].min(), df["InvoiceDate"].max())
```

```python
#orden por cantidad de menor a mayor
df.sort_values("Quantity").head(10)
```

```python
#orden por cantidad de mayor a menor 
df.sort_values("Quantity", ascending=False).head(10)
```

```python
#orden por precio unitario de menor a mayor 
df.sort_values("UnitPrice").head(10)
```

```python
#orden por precio unitario de mayor a menor
df.sort_values("UnitPrice", ascending=False).head(10)
```
